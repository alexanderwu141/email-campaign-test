### 1. Get the Hillstrom data

Downloads Kevin Hillstrom's MineThatData e-mail experiment (2008) using
`scikit-uplift`.

```python
import pandas as pd
import os
from sklift.datasets import fetch_hillstrom

ds = fetch_hillstrom(target_col="all")
```

### 2. Combine into one table and save

`fetch_hillstrom` returns three parts: customer features (`data`), the e-mail
group (`treatment`) and the outcomes (`target`). Put them back together in the
original column order and save it as `data/raw/hillstrom.csv`.

```python
df = ds.data.copy()
df["segment"] = ds.treatment
df = df.join(ds.target)

cols = ["recency", "history_segment", "history", "mens", "womens",
        "zip_code", "newbie", "channel", "segment",
        "visit", "conversion", "spend"]
df = df[cols]

os.makedirs("../data/raw", exist_ok=True)
df.to_csv("../data/raw/hillstrom.csv", index=False)
```
